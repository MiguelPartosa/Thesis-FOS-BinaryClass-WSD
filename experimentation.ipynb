# Splitting examples into groups
   - To be run in root

```python
import pandas as pd
import numpy as np

# Miguel set of first 36 Rows
df = pd.read_csv('./Dataset/Experimentation/UniqueVerbs.csv')
df = df.loc[0:35]
df
```

```python
# Shuffles the dataframe to avoid any bias when generating the sets 
df = df.sample(frac=1).reset_index(drop=True)

# Function to generate 15 rows per set (person)
def SplitExamples(df) -> list:
    example_set = []
    idx_example = 0 
    while len(example_set) < 5:
        # loop 15 times per set 
        verb_set = []
        for i in range(15):
            # Handle index out of range (36 rows)
            if idx_example > 35:
                idx_example = 0 
            verb_set.append(df.loc[idx_example].item())
            idx_example += 1
        example_set.append(pd.DataFrame({'Verbs': verb_set}))
    return example_set

first_36_df = SplitExamples(df)
for i in first_36_df:
    display(i)
```

## Save All Sets of Verbs
> You **MUST** change the `file_name` so you don't override the first files

```python
# For convenience hehe, Change this
file_name = 'First 36 Files'

file_name = file_name.replace(' ', '_')
for idx,i in enumerate(first_36_df):
    i.to_csv(f'Dataset/Experimentation/{file_name}{idx+1}.csv',index=False)
```
