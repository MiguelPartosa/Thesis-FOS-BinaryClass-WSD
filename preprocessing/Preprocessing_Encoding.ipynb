```python
from sentence_transformers import SentenceTransformer
model = SentenceTransformer('sentence-transformers/LaBSE')
```

```python
sentences = ["Daghan kog dulaan.", "Tigdula shag pusil."]
embeddings = model.encode(sentences)
```

```python
print(embeddings)
print("Similarity of embeddings: ",model.similarity(embeddings[0],embeddings[1]))
```
