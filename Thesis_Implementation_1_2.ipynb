## Word Sense

### Word Sense Cleaning

Removing tpye of FOS inside parenthesis


```python
df_verbs_filtered
```

```python
import re

# Example
sample_set = df_verbs_filtered.iloc[100:103, [2]]


def CleanWordSense(meaning: str) -> str:
    # Strip characters inside parenthesis inclusively
    clean = re.sub(r'\s*\([^)]*\)', '', meaning).strip()

    # Set of delimiters observed in the column
    delimiters = [';', '.', ',']

    pattern = f"[{''.join(re.escape(char) for char in delimiters)}]"

    # Find the first occurrence in clean
    match = re.search(pattern, clean)

    return clean[0:match.start()] if match else clean


sample_set['Meaning_en'].apply(CleanWordSense)
```

```python
df_cleaned_sense = df_verbs_filtered.copy()
df_cleaned_sense['Cleaned Word Sense'] = df_cleaned_sense.iloc[:, 2].apply(
    CleanWordSense)
df_cleaned_sense['Cleaned Word Sense']
```

```python
df_cleaned_sense.to_excel(
    '../../Dataset/Cleaned_Word_sense.xlsx', sheet_name="Cleaned Word Sense", index=False)
```

### Translate English Wordsense into Cebuano


```python
example = df_cleaned_sense.iloc[131:133].reset_index(drop=True)
example
```

Translator example from nllb


```python
def TranslateCebuano(text: str) -> str:
    try:
        translated_text = pipe(text, src_lang="eng_Latn", tgt_lang="ceb_Latn")
        return translated_text[0]['translation_text']
    except Exception as e:
        print(f"{e}\n\n Input:{text}")
        return ""
    
print("Original Example: ", example['Cleaned Word Sense'][0], "\n\nTranslated:", TranslateCebuano(
    example['Cleaned Word Sense'][0]))
```

```python
df_translated = df_cleaned_sense.copy()
df_translated["Translated Word Sense"] = df_cleaned_sense['Cleaned Word Sense'].apply(
    lambda x: TranslateCebuano(x))
df_translated
```

# Dataset Transformation

The following columns will be generated before the train and test split:

- (1) FOS/expression
- (2) their literal usage
- (3) their non-literal usage


Get the verb out of the given row


```python
import ast
# For converting the converted list strings into lists
def ExtractVerbs(tags: str | tuple) -> list:
    return [x[0].decode("utf-8") for x in ast.literal_eval(tags) if x[1] == 'VERB']
```

```python
df_translated['Verbs'] = df_translated['POS Tags'].apply(ExtractVerbs)
df_translated
```

```python
def FillCorrectVerbs(row):
    if not pd.isna(row['Corrected - Verbs']):
        row['Verbs'] = [row['Corrected - Verbs']]
    return row

df_translated = df_translated.apply(FillCorrectVerbs, axis=1)
df_translated
```

```python
df_translated['Figure of Speech'] = 1
df_translated.head(3)
```

### Train Test Split with Labels

```python
def SplitUsage(df_input:pd.DataFrame)-> pd.DataFrame:  
    df_split = pd.DataFrame()
    for x in df_input.iterrows():
        split_df = pd.DataFrame({
                'FOS': [x[1]['FOS'] for _ in range(2)],
                'Word Sense': [x[1]['Word Sense'] for _ in range(2)],
                'Verb': [x[1]['Verb'] for _ in range(2)],
                'Usage': [x[1]['Non-Literal Usage'], x[1]['Literal Usage']],
                'Is FOS' : [1,0]
                })
        
        df_split = pd.concat([df_split, split_df])
    return df_split.reset_index(drop=True)
    
SplitUsage(df_merged_final).head(3)
```

```python
df_merged_final_labels = SplitUsage(df_merged_final)
df_merged_final_labels
```
