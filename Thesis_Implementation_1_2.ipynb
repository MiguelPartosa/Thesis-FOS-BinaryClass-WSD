### Tag Verbs and Filter


```python
def ReturnTags(text: str) -> str:
    try:
        return tag_sentence(text)
    except Exception as e:
        print(f"{e}\n\n Input:{text}")
        return None
```

```python
df_filtered = df.copy()
df_filtered['POS Tags'] = df_filtered.iloc[:, 0].apply(lambda x: ReturnTags(x))
df_filtered
```

Translator example from nllb


```python
def TranslateCebuano(text: str) -> str:
    try:
        translated_text = pipe(text, src_lang="eng_Latn", tgt_lang="ceb_Latn")
        return translated_text[0]['translation_text']
    except Exception as e:
        print(f"{e}\n\n Input:{text}")
        return ""
```

```python
print("Original Example: ", example.iloc[:, 4][0], "\n\nTranslated:", TranslateCebuano(
    example.iloc[:, 4][0]))
```

```python
df_translated = df.copy()
df_translated.iloc[:, 4] = df_translated.iloc[:, 4].apply(
    lambda x: TranslateCebuano(x))
df_translated
```
